```python
import torch
from torch.utils.data import Dataset, TensorDataset, DataLoader
import torch.nn.functional as F
names = open("names.txt", "r").read().split()
alphabet = []
for name in names:
  for char in name:
    if char not in alphabet:
      alphabet.append(char)
alphabet = ["."] + sorted(alphabet)
print(alphabet)

charListLen = len(alphabet)
charToIndex = dict(zip(alphabet, range(charListLen)))
indexToChar = dict(zip(range(charListLen), alphabet))

inputLength = 3

inputs = []
label = []
for name in names:
  name = name + "."
  context = [0] * inputLength
  for char in name:
    answerIndex = charToIndex[char]
    inputs.append(context)
    label.append(answerIndex)
    context = context[1:]+[answerIndex]
print(len(names))

embeddingDimensions = 10
embeddingTable = torch.randn(charListLen, embeddingDimensions)

W1 = torch.randn(inputLength * embeddingDimensions, 100)
B1 = torch.randn(1, 100)

W2 = torch.randn(100, charListLen)
B2 = torch.randn(1, charListLen)
parameters = [embeddingTable, W1, B1, W2, B2]
for p in parameters:
  p.requires_grad=True
inputData = torch.tensor(inputs)
labelData = torch.tensor(label)
dataset = TensorDataset(inputData, labelData)
dataloader = DataLoader(dataset, batch_size = 32, shuffle = True)

epochs = 10
for i in range(epochs):
  totalLoss = 0
  index = 0
  for batchInput, batchLabel in dataloader:
    #print(batchInput.shape)
    embedding = embeddingTable[batchInput]
    embedding = embedding.view(embedding.shape[0], 30)
    x = embedding @ W1 + B1
    x=torch.tanh(x)
    x = x @ W2 + B2

    loss = F.cross_entropy(x, batchLabel)
    totalLoss += loss.item()
    index += 1
    if index%1000==0:
      print(loss.item())
    for p in parameters:
      p.grad = None
    loss.backward()

    for p in parameters:
      p.data -= p.grad * 0.01
  print(totalLoss/index)

  #x =

  #do funny stuff
```

```python
for _ in range(10):
  context = [0] * 3
  generatedName = []
  while True:
    embedding = embeddingTable[context].view(1, -1)
    x = embedding @ W1 + B1
    x = torch.tanh(x)
    x = x @ W2 + B2
    x = F.softmax(x, dim=1)
    nextIndex = torch.multinomial(x, num_samples=1).item()
    if nextIndex == 0:
      break
    generatedName.append(indexToChar[nextIndex])
    context = context[1:] + [nextIndex]
  print("".join(generatedName))
```
